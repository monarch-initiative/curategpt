Go get MatKG, convert to nodes/edges, index with curategpt

```python
kg_url = "http://kg-hub-public-data.s3.amazonaws.com/frozen_incoming_data/matkg-jun6.json"

# download the file
import requests
response = requests.get(kg_url)
# save the file
with open("matkg.json", "wb") as file:
    file.write(response.content)
```

```python
import json
import pandas as pd

# Load the JSON data
with open("matkg.json") as f:
    data = json.load(f)

# Get all unique keys across all "things"
all_keys = set()
for item in data["things"]:
    all_keys.update(item.keys())

# Sort keys for consistent column order
all_keys = sorted(all_keys)

# Build rows with all fields, filling in missing ones
rows = []
for item in data["things"]:
    row = {}
    for key in all_keys:
        value = item.get(key, "")
        # Convert lists and dicts to JSON strings
        if isinstance(value, (list, dict)):
            value = json.dumps(value)
        row[key] = value
    rows.append(row)

# Create DataFrame and export to TSV
df = pd.DataFrame(rows)
df.to_csv("things.tsv", sep="\t", index=False)

# Display the DataFrame
df.head()
```

```python
# add column model to things.tsv, with value of "all-MiniLM-L6-v2"
df["model"] = "all-MiniLM-L6-v2"
df.to_csv("things.tsv", sep="\t", index=False)
```
